## Problem 1: Sum values in even-numbered positions

Positions are counted starting at 1. Therefore, the values in positions 2, 4, 6, and so on are included.


### Python solution


```python
from numbers import Number


def sum_even_positions(values):
    """Return the sum of values in positions 2, 4, 6, ...

    Args:
        values: A sequence containing numeric values.

    Returns:
        The sum of the values in even-numbered positions.

    Raises:
        TypeError: If a selected value is not numeric.
    """
    selected_values = values[1::2]

    if not all(isinstance(value, Number) for value in selected_values):
        raise TypeError("All selected values must be numeric.")

    return sum(selected_values)
```

### Python examples


```python
examples = [
    [3, 5, 1, 7],
    [10, 20, 30, 40],
    [],
    [9],
]

for values in examples:
    print(f"{values} -> {sum_even_positions(values)}")
```

### Complexity

- Time complexity: **O(n)**
- Space complexity: **O(n)** for the sliced list

A generator-based implementation could reduce the additional space to **O(1)**.


## Problem 2: Two Sum

Given a list of integers and a target, return the indices of two distinct values whose sum equals the target. Python indices are zero-based.


### Efficient Python solution


```python
def two_sum(nums, target):
    """Return indices of two values that add up to target.

    Args:
        nums: A sequence of numeric values.
        target: Desired sum.

    Returns:
        A two-element list containing zero-based indices.

    Raises:
        ValueError: If no valid pair exists.
    """
    seen = {}

    for index, value in enumerate(nums):
        complement = target - value

        if complement in seen:
            return [seen[complement], index]

        seen[value] = index

    raise ValueError("No two values add up to the target.")
```

### Python examples


```python
test_cases = [
    ([2, 1, 7, 23], 8),
    ([3, 4, 12, 7], 10),
    ([3, 3], 6),
]

for nums, target in test_cases:
    print(f"nums={nums}, target={target} -> indices={two_sum(nums, target)}")
```

### Python validation tests


```python
assert sum_even_positions([3, 5, 1, 7]) == 12
assert sum_even_positions([10, 20, 30, 40]) == 60
assert sum_even_positions([]) == 0
assert sum_even_positions([9]) == 0

assert two_sum([2, 1, 7, 23], 8) == [1, 2]
assert two_sum([3, 4, 12, 7], 10) == [0, 3]
assert two_sum([3, 3], 6) == [0, 1]

try:
    two_sum([1, 2, 3], 100)
except ValueError:
    pass
else:
    raise AssertionError("Expected ValueError when no solution exists.")

print("All Python tests passed.")
```

### Complexity

The optimized `two_sum` solution stores previously seen values in a dictionary.

- Time complexity: **O(n)**
- Space complexity: **O(n)**

This improves on the original nested-loop approach, which has **O(n²)** time complexity.


## R implementations

The following cells require R and the `rpy2` Jupyter extension. R uses one-based indexing, so the returned positions differ from Python's zero-based indices.


```python
%load_ext rpy2.ipython
```

### R solution: sum values in even-numbered positions


```python
%%R

sum_even_positions <- function(values) {
  if (length(values) < 2) {
    return(0)
  }

  even_positions <- seq(2, length(values), by = 2)
  sum(values[even_positions])
}
```

```python
%%R

examples <- list(
  c(3, 5, 1, 7),
  c(10, 20, 30, 40),
  numeric(0),
  c(9)
)

sapply(examples, sum_even_positions)
```

### R solution: Two Sum


```python
%%R

two_sum <- function(nums, target) {
  seen <- new.env(hash = TRUE, parent = emptyenv())

  for (i in seq_along(nums)) {
    complement <- target - nums[i]
    key <- as.character(complement)

    if (exists(key, envir = seen, inherits = FALSE)) {
      return(c(get(key, envir = seen), i))
    }

    assign(as.character(nums[i]), i, envir = seen)
  }

  stop("No two values add up to the target.")
}
```

```python
%%R

two_sum(c(2, 1, 7, 23), 8)
two_sum(c(3, 4, 12, 7), 10)
two_sum(c(3, 3), 6)
```
