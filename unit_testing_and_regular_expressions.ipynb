## 1. Unit testing


```python
import unittest
```

### Calculator functions


```python
def add(x, y):
    """Return the sum of two numbers."""
    return x + y


def subtract(x, y):
    """Return the difference between two numbers."""
    return x - y


def multiply(x, y):
    """Return the product of two numbers."""
    return x * y


def divide(x, y):
    """Return x divided by y.

    Raises:
        ZeroDivisionError: If y is zero.
    """
    if y == 0:
        raise ZeroDivisionError("Cannot divide by zero.")
    return x / y
```

### Test suite


```python
class TestCalculator(unittest.TestCase):
    def test_addition(self):
        self.assertEqual(add(6, 4), 10)
        self.assertEqual(add(-4, -6), -10)
        self.assertEqual(add(4, -2), 2)

    def test_subtraction(self):
        self.assertEqual(subtract(7, 3), 4)
        self.assertEqual(subtract(-8, -2), -6)
        self.assertEqual(subtract(4, -6), 10)

    def test_multiplication(self):
        self.assertEqual(multiply(3, 4), 12)
        self.assertEqual(multiply(-2, -3), 6)
        self.assertEqual(multiply(4, -3), -12)

    def test_division(self):
        self.assertEqual(divide(12, 3), 4)
        self.assertEqual(divide(-12, -3), 4)
        self.assertEqual(divide(12, -3), -4)

    def test_division_by_zero(self):
        with self.assertRaises(ZeroDivisionError):
            divide(10, 0)
```

### Run the tests


```python
test_result = unittest.main(argv=[""], verbosity=2, exit=False)
```

## 2. Regular expressions


```python
import re
```

### Sample text


```python
text = (
    "To make an appointment, please email me at "
    "marryg@gmail.com or marryg@harrisburgU.edu."
)

text
```

### Search for a word


```python
search_term = "please"
match = re.search(rf"\b{re.escape(search_term)}\b", text, flags=re.IGNORECASE)

if match:
    print(f"Found '{search_term}'.")
else:
    print(f"Did not find '{search_term}'.")
```

### Extract email addresses


```python
email_pattern = re.compile(
    r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"
)

emails = email_pattern.findall(text)
emails
```

## Conclusion

The unit tests verify calculator behavior across positive, negative, mixed-sign, and zero-division cases. The regular-expression examples show how to locate a word and extract email addresses from unstructured text.
