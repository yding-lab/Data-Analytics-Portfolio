# Data Wrangling with pandas

This notebook demonstrates foundational data-wrangling techniques using a small veterinary-visit dataset. It covers DataFrame creation, inspection, filtering, missing-value handling, sorting, aggregation, grouping, and pivot tables.

## Skills demonstrated

- Creating a pandas DataFrame
- Selecting rows and columns with `loc` and `iloc`
- Filtering with Boolean conditions
- Handling missing values
- Updating, adding, and deleting rows
- Aggregating and grouping data
- Sorting values
- Creating pivot tables


## 1. Setup


```python
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
```

## 2. Create the sample dataset


```python
data = {
    "animal": ["cat", "cat", "snake", "dog", "dog", "cat", "snake", "cat", "dog", "dog"],
    "age": [2.5, 3, 0.5, np.nan, 5, 2, 4.5, np.nan, 7, 3],
    "visits": [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    "priority": ["yes", "yes", "no", "yes", "no", "no", "no", "yes", "no", "no"],
}

labels = list("abcdefghij")

df = pd.DataFrame(data, index=labels)
df
```

## 3. Inspect the DataFrame


### Basic structure and data types


```python
df.info()
```

### First three rows


```python
df.head(3)
```

## 4. Select rows and columns


### Select the `animal` and `age` columns


```python
df[["animal", "age"]]
```

### Select rows at positions 3, 4, and 8 and display `animal` and `age`


```python
df.iloc[[3, 4, 8]][["animal", "age"]]
```

## 5. Filter records


### Records with more than three visits


```python
df.loc[df["visits"] > 3]
```

> The result is empty because no record in this sample has more than three visits.


### Records with missing ages


```python
df.loc[df["age"].isna()]
```

### Cats younger than three


```python
df.loc[(df["animal"] == "cat") & (df["age"] < 3)]
```

### Records with ages between two and four, inclusive


```python
df.loc[df["age"].between(2, 4, inclusive="both")]
```

## 6. Update the data


### Change the age in row `f` to 1.5


```python
df.loc["f", "age"] = 1.5
df
```

## 7. Aggregate and summarize


### Total number of visits


```python
total_visits = df["visits"].sum()
total_visits
```

### Mean age by animal type


```python
mean_age_by_animal = df.groupby("animal", dropna=False)["age"].mean()
mean_age_by_animal
```

## 8. Add and remove a row


```python
df.loc["k"] = ["dog", 4.0, 2, "yes"]
df_with_new_row = df.copy()

df = df.drop(index="k")

print("DataFrame after adding row k:")
display(df_with_new_row)

print("DataFrame after removing row k:")
display(df)
```

## 9. Count and sort records


### Number of records for each animal type


```python
df["animal"].value_counts()
```

### Sort by age descending and visits ascending


```python
df.sort_values(by=["age", "visits"], ascending=[False, True], na_position="last")
```

## 10. Create a pivot table


```python
age_pivot = df.pivot_table(
    index="animal",
    columns="visits",
    values="age",
    aggfunc="mean",
)

age_pivot
```
